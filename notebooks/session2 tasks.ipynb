```python
#List statistics​
#Ask for numbers until the user types done. Keep them in a list, then print the count, mean, smallest and largest — and, with a comprehension, the numbers above the mean.​
nums = []
while True:
    numbers = input("enter numbers (or done): ").strip()
    if numbers == "done":
        break
    nums.append(float(numbers))
x = sum(nums) / len(nums)
y = [i for i in nums if i > x]
print(f"Count: {len(nums)}  Mean: {x:.2f}")
print(f"Min: {min(nums)}  Max: {max(nums)}")
print("Above the mean:", y)
```

```python
#Word counter from a file​
#Write the paragraph given in the solution to story.txt, read it back, remove punctuation, count the words with a dictionary and print the five most common.​
from pathlib import Path
Path("story.txt").write_text(
    "The quick brown fox jumps over the lazy dog. "
    "The dog sleeps; the fox runs.", encoding="utf-8")
text = Path("story.txt").read_text(encoding="utf-8").lower()
for panc in ".,;!?":
    text = text.replace(panc, "")
counts = {}
for words in text.split():
    counts[words] = counts.get(words,0)+1
for words in sorted(counts, key=counts.get, reverse=True)[:5]:
    print(words,counts[words])
```

```python
# JSON contact book
import json
from pathlib import Path

def contacts(university, name, phone):
    university[name] = phone

def find(university, name):
    return university.get(name, "not in university")

def save(university, path):
    text = json.dumps(university, indent=2)
    Path(path).write_text(text, encoding="utf-8")

def load(path):
    text = Path(path).read_text(encoding="utf-8")
    return json.loads(text)

university = {}
contacts(university, "Mona", "0100 123 4567")
contacts(university, "Omar", "0122 765 4321")
save(university, "contacts.json")

loaded = load("contacts.json")
print(find(loaded, "Omar"))
print(find(loaded, "Sara"))
print(loaded == university)
```

```python
#Safe calculator​
#Ask for two numbers and an operator (+ − × ÷). Keep asking until each number is valid; handle ValueError and ZeroDivisionError with clear messages.​
def read_number(prompt):
    while True:
        try:
            return float(input(prompt))
        except ValueError:
            print("Not a number, try again.")

a = read_number("First: ")
b = read_number("Second: ")
op = input("Operator (+ - * /): ").strip()

try:
    if op == "+":
        result = a + b
    elif op == "-":
        result = a - b
    elif op == "*":
        result = a * b
    elif op == "/":
        result = a / b
    else:
        raise ValueError(f"unknown operator {op}")
    print("Result:", result)
except ZeroDivisionError:
    print("Cannot divide by zero.")
except ValueError as error:
    print("Error:", error)
```

```python
#Bank account with history​
#Extend BankAccount so each deposit and withdrawal is stored as a (kind, amount) tuple in self.history, and add statement() to print the history and the balance.​
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
        self.history = []

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("deposit must be positive")
        self.balance += amount
        self.history.append(("deposit", amount))

    def withdraw(self, amount):
        if amount > self.balance:
            raise ValueError(f"only {self.balance} available")
        self.balance -= amount
        self.history.append(("withdraw", amount))

    def statement(self):
        for kind, amount in self.history:
            print(f"{kind:<9}{amount:>6}")
        print(f"{'balance':<9}{self.balance:>6}")

    def __str__(self):
        return f"{self.owner}: {self.balance}"

account = BankAccount("Mona", 100)
account.deposit(50)
account.withdraw(30)
account.statement()
```

```python
%%writefile gradebook.py
def average(scores):
    return sum(scores) / len(scores)

def letter(avg):
    if avg >= 90:
        return "A"
    if avg >= 80:
        return "B"
    if avg >= 70:
        return "C"
    return "F"
```

```python
import json
import gradebook

students = {"Mona": [92, 88, 95],
            "Omar": [75, 68, 80],
            "Sara": [85, 90, 78]}

report = {}
for name, scores in students.items():
    avg = gradebook.average(scores)
    report[name] = {"average": round(avg, 1),
                    "grade": gradebook.letter(avg)}

with open("report.json", "w", encoding="utf-8") as f:
    json.dump(report, f, indent=2)

print(json.dumps(report["Omar"]))
```
