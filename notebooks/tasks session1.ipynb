```python
#Temperature converter​ Ask for a temperature in Celsius. Print it in Fahrenheit (one decimal) and Kelvin (two decimals). F = C × 9 / 5 + 32 · K = C + 273.15​
h = float(input("enter the temp in celsius: "))
g = h * 9 / 5 + 32
d= h+273.15
print(f"Fahrenheit: {g:.1f}")
print(f"kelvin: {d:.2f}")
```

```python
#Even and odd counter​
#Ask for a whole number n. Using a loop — not a formula — count how many numbers from 1 to n are even and how many are odd. Print both counts.​
n = int(input("enter a whole num: "))
even = 0
odd = 0
for i in range(1,n+1):
    if i%2==0:
        even +=1
    else:
        odd+=1
print(f"even num: {even}, odd num: {odd}")
```

```python
#Leap-year function​
#Write is_leap(year) returning True or False. Leap years divide by 4, except those dividing by 100 — unless they also divide by 400. Test 2024, 1900, 2000, 2026.​
def is_leep(year):
    if year %400==0:
        return True
    if year %100==0:
        return False
    return year %4==0
for i in (2024,1900,2000,2026):
    print (i, is_leep(i))
```

```python
#Prime checker​
#Write is_prime(n): True if n is greater than 1 and divisible only by 1 and itself. Loop over possible divisors and stop early when you find one. Print every prime below 50.​
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2,int( n **0.5)+1):
        if n %i==0:
            return False
    return True
for n in range(2,50):
    if is_prime(n):
        print(n,end="") 
```

```python
#Menu calculator​
#In a loop, show a menu (1 add, 2 subtract, 3 multiply, 4 divide, 5 quit). Read the choice and two numbers, print the result, repeat until 5. Put the maths in calculate(choice, a, b) and refuse division by zero.​
def calc(choose, x,y):
    if choose == 1:
        return x+y
    if choose==2:
        return x-y
    if choose==3:
        return x*y
    if choose==4:
        if y==0:
            return "couldn't devided by 0"
        return x/y
while True:
    choose = int(input("1+ 2- 3* 4/ 5=>quit : "))
    if choose==5:
        print("try again..:)")
        break
    if choose not in (1,2,3,4,):
        print ("invalid mum :)")
        continue
    x = float(input("first num: "))
    y = float(input("second num: "))    
    print("result: ",calc(choose, x,y))
```

```python
#Guessing game+​
#Extend the build-along: at most 7 tries, show how many are left, and ask "Play again? (y/n)" at the end. Stretch: let the player choose the range.​
import random 
tries= 7
def play(low,high):
    x=random.randint(low,high)
    for i in range(1,tries+1):
        left = tries - i +1
        guess = int(input(f"guess ({low}-{high}),{left} tries left: "))
        if guess == x:
            print (f"great :) {x} in {tries} tries.")
            return
        elif guess < x:
            print("too low")
        else:
            print("too high")
    print(f"out of tries. the num was {x}")
while True:
    low = int(input("lower num: "))
    high = int(input("higher num: "))
    play(low,high)
    again = input("try again..? (y/n) ").strip().lower()
    if again != "y":
        print("see you soon :)")
        break

```
